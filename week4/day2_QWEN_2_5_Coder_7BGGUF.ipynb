```python
from huggingface_hub import hf_hub_download
from llama_cpp import Llama
```

```python
# ✅ Choose model repo and filename
repo_id = "Qwen/Qwen2.5-Coder-7B-Instruct-GGUF"
filename = "qwen2.5-coder-7b-instruct-q5_k_m.gguf"  # or q4_k_m, q3_k_m

# ✅ Download GGUF model to local cache (automatically)
model_path = hf_hub_download(repo_id=repo_id, filename=filename)
print(f"Model downloaded to: {model_path}")
```

```python
# ✅ Load model with llama-cpp
llm = Llama(
    model_path=model_path,
    n_ctx=2048,
    n_threads=4  # Adjust to your CPU cores
)
```

```python
# ✅ Prompt the model
prompt = "### Human: Write a Python function to reverse a string.\n### Assistant:"
output = llm(prompt, max_tokens=1000)
```

```python
# ✅ Display output
print(output["choices"][0]["text"])
```
