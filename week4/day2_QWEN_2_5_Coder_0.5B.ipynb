```python
from transformers import AutoModelForCausalLM, AutoTokenizer
#import torch
```

```python
print ("Loading model...")
model_name = "Qwen/Qwen2.5-0.5B-Instruct"  # Much faster and works on CPU

# Load tokenizer and quantized model (auto uses GPU if available)
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(model_name, trust_remote_code=True).to("cpu")
```

```python
print ("Generating inputs...")
inputs = tokenizer("Write a function to reverse a string in Python:", return_tensors="pt")
```

```python
print ("Generating outputs...")
outputs = model.generate(**inputs, max_new_tokens=2000)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
